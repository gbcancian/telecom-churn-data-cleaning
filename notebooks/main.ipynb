## Normalizando colunas com formato JSON

```python
with open('../data/dataset-telecon.json') as f:
    json_raw = json.load(f)
    
data_normalized = pd.json_normalize(json_raw)
data_normalized.head()
```

```python
data_normalized.info()
```

```python
#data_normalized['conta.cobranca.Total'] = data_normalized['conta.cobranca.Total'].astype(float){}
#data_normalized[data_normalized['conta.cobranca.Total'] == ' '].head()
data_normalized['conta.cobranca.Total'] = data_normalized['conta.cobranca.Total'].astype(object)
data_normalized[data_normalized['conta.cobranca.Total'] == ' '][['cliente.tempo_servico','conta.contrato','conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
idx = data_normalized[data_normalized['conta.cobranca.Total'] == ' '].index
data_normalized.loc[idx, 'cliente.tempo_servico'] = 24
data_normalized.loc[idx, 'conta.cobranca.Total'] = data_normalized.loc[idx,'cliente.tempo_servico'] * data_normalized.loc[idx,'conta.cobranca.mensal']
data_normalized.loc[idx, ['cliente.tempo_servico','conta.contrato','conta.cobranca.mensal', 'conta.cobranca.Total']]
                                      
```

```python
data_normalized['conta.cobranca.Total'] = data_normalized['conta.cobranca.Total'].astype(float)
```

# Identificando e tratando strings vazias

# Identificando e tratando dados duplicados

```python
#Verificando duplicatas
data_normalized.duplicated().sum()
```

```python
#Criando um filtro para verificar duplicatas
filter_duplicated = data_normalized.duplicated()
data_normalized[filter_duplicated].head()
```

```python
#Removendo duplicatas
data_normalized.drop_duplicates(inplace=True)
data_normalized.duplicated().sum()
```

# Identificando e tratando dados nulos

```python
data_normalized.isna().sum()
```

```python
data_normalized[data_normalized['cliente.tempo_servico'].isna()][['cliente.tempo_servico','conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
idx = data_normalized[data_normalized['cliente.tempo_servico'].isna()].index
data_normalized.loc[idx, 'cliente.tempo_servico'] = data_normalized.loc[idx, 'conta.cobranca.Total'] / data_normalized.loc[idx, 'conta.cobranca.mensal'].round()
data_normalized['cliente.tempo_servico'].isna().sum()
```

```python
filter_drop = ['conta.contrato','conta.faturamente_eletronico','conta.metodo_pagamento']
data_normalized[filter_drop].isna().any(axis=1).sum()
```

```python
data_normalized = data_normalized.dropna(subset=filter_drop).copy()
data_normalized.reset_index(drop=True, inplace=True)
data_normalized.isna().sum()
```

# Variáveis categóricas

```python
data_normalized.head()
```

```python
#Removendo dados de Id do cliente, já que são dispensáveis para dados de machine learning
data_normalized = data_normalized.drop('id_cliente', axis=1)
data_normalized.head()
```

```python
#Consultando variavéis categoricas binárias
for col in data_normalized.columns:
    print(f'Coluna: {col}')
    print(data_normalized[col].unique())
    print('-'*30)
```

```python
#Realizando a substituição para dados binários
mapeamento = {
    'nao': 0,
    'sim': 1,
    'masculino': 0,
    'feminino': 1,
}
colunas = ['Churn', 'cliente.genero', 'cliente.parceiro', 'cliente.dependentes', 'telefone.servico_telefone', 'conta.faturamente_eletronico']

data_normalized[colunas] = data_normalized[colunas].replace(mapeamento)
data_normalized
```
